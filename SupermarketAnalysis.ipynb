### สร้างฐานข้อมูล SQLite

```
# เรามีข้อมูลยอดขายอยู่ใน Python ชื่อ df_clean
# และต้องการนำข้อมูลทั้งหมดไปเก็บในฐานข้อมูล SQLite ชื่อ supermarket.db
# โดยสร้างตารางชื่อ sales เพื่อให้เราสามารถเขียน SQL วิเคราะห์ข้อมูลได้

# import คือการนำ Library หรือ Module มาใช้งาน
# sqlite3 เป็น Module ที่มีมาให้ใน Python
# ใช้สำหรับเชื่อมต่อและจัดการฐานข้อมูล SQLite

import sqlite3

# sqlite3.connect() ใช้เปิดการเชื่อมต่อกับฐานข้อมูล
# 'supermarket.db' คือชื่อไฟล์ฐานข้อมูล
# ถ้าไฟล์ยังไม่มี SQLite จะสร้างไฟล์ใหม่
# conn คือตัวแปรที่เก็บการเชื่อมต่อนี้

conn = sqlite3.connect('supermarket.db')

# supermarket.db = อาคารที่เก็บข้อมูล
# sales = ห้องหนึ่งในอาคาร
# conn = กุญแจที่ช่วยให้ Python เข้าไปจัดการข้อมูลในอาคาร

# นำข้อมูลจาก df_clean ไปเก็บเป็นตาราง SQL
df_clean.to_sql('sales',conn,if_exists='replace',index=False)

# df_clean	DataFrame ที่เก็บข้อมูลยอดขายที่เตรียมไว้
# .to_sql()	ฟังก์ชันของ pandas สำหรับเขียนข้อมูลลงฐานข้อมูล SQL
# 'sales'	ชื่อตารางที่เราต้องการสร้าง
# conn	การเชื่อมต่อฐานข้อมูลที่สร้างไว้
# if_exists='replace'	ถ้าตารางเดิมมีอยู่ ให้ลบแล้วสร้างใหม่จาก DataFrame
# index=False	ไม่บันทึก Index ของ DataFrame เป็นคอลัมน์เพิ่มเติม

# Index 0,1,2 python เลขตรงนั้นเรียกว่า DataFrame Index ซึ่งเป็นป้ายกำกับแถวของ pandas
# เมื่อกำหนด index=False ฐานข้อมูลจะเก็บเฉพาะคอลัมน์ Branch และ Total ไม่เพิ่มคอลัมน์ Index เข้าไปอีก
# ข้อมูลของเรา มี Invoice ID ที่ใช้ระบุธุรกรรมอยู่แล้ว จึงไม่จำเป็นต้องนำ Index ของ pandas ไปเก็บเพิ่ม
```

### ทดสอบเขียน SQL Query
ดึงข้อมูลยอดขายรวมแยกตามสาขา (Branch)

```
# -- เลือกสาขาและเมือง
# -- SUM(Total) รวมยอดขายของแต่ละกลุ่ม
# -- AS Total_Sales ตั้งชื่อคอลัมน์ผลลัพธ์ใหม่
# -- COUNT(*) นับจำนวนแถวในแต่ละกลุ่ม
# SELECT Branch, City, SUM(Total) AS Total_Sales, COUNT(*) AS Transaction_Count

#-- เลือกข้อมูลจากตาราง sales FROM sales

#-- จัดกลุ่มข้อมูลตามสาขาและเมือง GROUP BY Branch, City

#-- เรียงจากยอดขายสูงสุดไปต่ำสุด ORDER BY Total_Sales DESC;
#-- เรียงจากยอดขายต่ำสุดไปสูงสุด ORDER BY Total_Sales ASC;
```

```
# เขียนคำสั่ง SQL เพื่อจัดกลุ่มและหาผลรวมยอดขายตามสาขา (Branch) และเมือง (City)
query_analysis = """
SELECT Branch, City, SUM(Total) AS Total_Sales, COUNT(*) AS Transaction_Count
FROM sales
GROUP BY Branch, City
ORDER BY Total_Sales DESC;
"""

# ส่ง SQL ไปประมวลผลผ่าน conn
analysis_result = pd.read_sql_query(query_analysis, conn)

# แสดงผลลัพธ์
display(analysis_result)
```

### วิเคราะห์ยอดขายแยกตามประเภทสินค้า (Product Line)
เขียน SQL Query เพื่อสรุปยอดขายรวมของแต่ละ Product Line โดยเรียงจากยอดขายสูงที่สุดลงไป

```
# เขียนคำสั่ง SQL เพื่อหาผลรวมยอดขายแยกตามประเภทสินค้า (Product line)
query_product_sales = """
SELECT `Product line`, SUM(Total) AS Total_Sales, COUNT(*) AS Transaction_Count
FROM sales
GROUP BY `Product line`
ORDER BY Total_Sales DESC;
"""

# ดึงข้อมูลผ่าน pandas
product_sales_result = pd.read_sql_query(query_product_sales, conn)

# แสดงผลลัพธ์
display(product_sales_result)
```

```
# pd.to_datetime() แปลงข้อมูลเวลาให้เป็นชนิด datetime
# df_clean['Time'] คือคอลัมน์ที่เราต้องการแปลง
# format='%H:%M:%S' บอกว่าเวลาอยู่ในรูป ชั่วโมง:นาที:วินาที
# %H = ชั่วโมงแบบ 24 ชั่วโมง (00–23)
# %M = นาที (00–59)
# %S = วินาที (00–59)
# errors='coerce' ให้ค่าที่แปลงไม่ได้กลายเป็น NaT

df_clean['Time'] = pd.to_datetime(
    df_clean['Time'],
    format='%H:%M:%S',
    errors='coerce')

# .dt.hour ดึงเฉพาะชั่วโมงออกจากข้อมูล datetime
# แล้วสร้างคอลัมน์ใหม่ชื่อ Hour
df_clean['Hour'] = df_clean['Time'].dt.hour

# แสดง Time และ Hour จำนวน 10 แถวแรก
display(df_clean[['Time', 'Hour']].head(10))
```

```
# นำ DataFrame ที่มีคอลัมน์ Hour ไปบันทึกใน SQLite
# replace = ลบตาราง sales เดิมและสร้างใหม่จาก df_clean
# index=False = ไม่บันทึกเลข Index ของ pandas
df_clean.to_sql('sales', conn, if_exists='replace', index=False)

# supermarket.db คือแฟ้มเอกสาร
# ส่วน sales คือชีตข้อมูลหนึ่งชีตในแฟ้มนั้น
# และ df_clean คือตารางที่เราเตรียมใน Python ก่อนนำไปเก็บในแฟ้ม

# แสดงข้อความหลังบันทึกสำเร็จ
print("Database updated!")
```

```
# สร้างตัวแปรเพื่อเก็บคำสั่ง SQL
query_hourly = """

-- SELECT เลือกข้อมูลที่ต้องการแสดง
SELECT
    Hour,

    -- COUNT(*) นับจำนวนธุรกรรมในแต่ละชั่วโมง
    COUNT(*) AS Transaction_Count

-- FROM เลือกตาราง sales ใน SQLite
FROM sales

-- GROUP BY จัดข้อมูลที่มีชั่วโมงเดียวกันเป็นกลุ่ม
GROUP BY Hour

-- ORDER BY เรียงชั่วโมงจากน้อยไปมาก
-- ASC = Ascending (น้อยไปมาก)
ORDER BY Hour ASC;

"""

# ใช้ Python ส่ง SQL ไปประมวลผลใน SQLite
# แล้วเก็บผลลัพธ์เป็น DataFrame ชื่อ hourly_result
hourly_result = pd.read_sql_query(query_hourly, conn)

# แสดงผลตาราง
display(hourly_result)
```
